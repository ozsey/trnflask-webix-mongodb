**PROGRAM 2**

```python
n = int(input("Input tinggi piramida(n): "))

for i in range(n):
  for j in range(i+1):
    print(' ',end=' ')
  for k in range(n-i):
    print('\U0001F601 ',end=' ')
  print()
```

```python
from ast import LShift
LH = [10000, 15000, 5000, 20000]
Br = ["Beras", "Telur", "Minyak", "Terigu"]
disc = [0.2, 0.07, 0.03, 0.8]

print("[ PROGRAM DISKON BARANG ]")
print("[1] Beras: 70000,\tDisc: 20%\n[2] Telur: 45000,\tDisc: 7%\n[3] Minyak: 20000,\tDisc: 3%\n[4] Terigu: 17000,\tDisc: 8%\n[E] Exit\n - - -")

while True:
  pilihan = input("\nInput user: ")

  if pilihan.upper() == "E":
    break

  elif int(pilihan) >= 1 and int(pilihan) <= 4:
    ad = LH[int(pilihan) - 1] - (LH[int(pilihan) - 1] * disc[int(pilihan) - 1])
    print("Nama barang: ", Br[int(pilihan) - 1])
    print("Harga barang: ", LH[int(pilihan) - 1])
    print("Harga setelah diskon: ", ad)

  else:
    print("Barang tidak tersedia.")
```
