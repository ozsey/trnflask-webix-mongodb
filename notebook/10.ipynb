```python
number1 = {
    'a': [1, 2, 3, 4, 5, 6],
    'b': [15, 16, 17, 18],
    'c': 17,
    'd': 18392,
    'e': [1372, 43347, 717162, 1223],
    'f': 20
}

def number1_func(inputList):
  result = {
        key: value[::-1] if isinstance(value, list) else value for key, value in inputList.items()
      }
  return result

print(number1_func(number1))
```

```python
tuple1 = (10, 20, (20, 10), 30, (40, 50, 60), 60, 70)

freq = {}

for item in tuple1:
  if isinstance(item, tuple):
    for subitem in item:
      if subitem in freq:
        freq[subitem] += 1
      else:
        freq[subitem] = 1
  else:
    if item in freq:
      freq[item] += 1
    else:
      freq[item] = 1

print(freq)
```

```python
# Soal Tuple
tuple1 = (10, 20, (20, 10), 30, (40, 50, 60), 60, 70)

#buatkan program untuk menghitung frekuensi elemen dalam tuple1 (dalam dictionary)
#output : {10: 2, 20: 2, 30: 1, 40: 1, 50: 1, 60: 2, 70: 1}

tuple_to_list = list(tuple1)

index2 = list(tuple_to_list[2])
del(tuple_to_list[2])
index3 = list(tuple_to_list[3])
del(tuple_to_list[3])

get_list = tuple_to_list + index2 + index3

sorted_list = sorted(get_list)
print("sorted_list:",sorted_list)

result ={}
for item in sorted_list:
  if item in result:
    result[item] += 1
  else:
    result[item] = 1

print("final:",result)
```
