```python
#Urutkan Dictionary
test_dict = {
    'a': [7, 6, 3],
    'g': [2, 10, 3],
    'b': [19, 4],
    'h': [1, 5, 8, 12],
    'd': [15, 20],
    'e': [9, 11, 13, 17],
    'i': [14, 16, 18],
    'c': [21, 22, 23],
    'f': [24, 25]
}

#Output nya ada 2 yaitu urutkan dari yang terkecil dan jumlahkan per keynya
# {'a': [3, 6, 7], 'b': [4, 19], 'c': [21, 22, 23], 'd': [15, 20], 'e': [9, 11, 13, 17], 'f': [24, 25], 'g': [2, 3, 10], 'h': [1, 5, 8, 12], 'i': [14, 16, 18]}
# {'a': 16, 'g': 15, 'b': 23, 'h': 26, 'd': 35, 'e': 50, 'i': 48, 'c': 66, 'f': 49}


def soalDictionary(inputDict):
  hasil_sort = {}
  for key, value in inputDict.items():
    value.sort()
    hasil_sort.update({key: value})

  hasil_sum = {}
  for key, value in inputDict.items():
    summed = sum(value)
    hasil_sum.update({key: summed})

  print(hasil_sort, hasil_sum, sep='\n')

soalDictionary(test_dict)
```

```python
# Soal Tuple
tuple1 = (10, 20, (20, 10), 30, (40, 50, 60), 60, 70)

#buatkan program untuk menghitung frekuensi elemen dalam tuple1 (dalam dictionary)
#output : {10: 2, 20: 2, 30: 1, 40: 1, 50: 1, 60: 2, 70: 1}

tuple_to_list = list(tuple1)

index2 = list(tuple_to_list[2])
del(tuple_to_list[2])
index3 = list(tuple_to_list[3])
del(tuple_to_list[3])

get_list = tuple_to_list + index2 + index3

sorted_list = sorted(get_list)
print(sorted_list)


result = {}
for item in sorted_list:
  if item in result:
    result[item] += 1
  else:
    result[item] = 1

print(result)
```

```python
my_tuple = (10, 20, 30, 40, 50)

# Soal: Hitung rata-rata dari semua elemen di dalam tuple

# Solusi:
sum = 0
for item in my_tuple:
  sum += item

average = sum / len(my_tuple)

print("Rata-rata elemen tuple:", average)
```

```python
number1 = {
    'a': [1, 2, 3, 4, 5, 6],
    'b': [15, 16, 17, 18],
    'c': 17,
    'd': 18392,
    'e': [1372, 43347, 717162, 1223],
    'f': 20
}

def number1_func():
  result = {}
  for key, val in number1.items():
    if isinstance(val, list):
      result[key] = val[(len(val)):None:-1]
    else:
      result[key] = val
  return result

print(number1_func())
```
