# Number 1
Sesuaikan dengan kategorinya dan hapus angka dalam dictionary-nya! Jadikan 1 dictionary!


SOLUSI:

```python
# sesuaikan dengan kategorinya dan hapus angka dalam dict nya
x = {'buah':('apel', 'anggur', 'bayam'),'1':(1,2,3), 'sayur':('pisang', 'wortel')}
y = {'daging':('ayam', 'sapi')}
temp = x.pop('1')
buah = list(x.get('buah'))
buah.pop(2)
buah.append("pisang")
sayur = list(x.get('sayur'))
sayur.pop(0)
sayur.append("bayam")

hasil = {"buah": buah, "sayur": sayur }
hasil.update(y)
print(hasil)
```

# Number 3
Terdapat sebuah dictionary dengan value berupa integer dan list integer. Untuk masing-masing value yang terdapat di dalam value dictionary, reverse list yang terdapat dalam value tersebut, termasuk integer yang ada.



```python
number1 = {
    'a': [1, 2, 3, 4, 5, 6],
    'b': [15, 16, 17, 18],
    'c': 17,
    'd': 18392,
    'e': [1372, 43347, 717162, 1223],
    'f': 20
}

'''
contoh hasil:
{
    'b': [18, 17, 16, 15],
    'c': 17 #tetap sama, int
}
'''

def number1_func():
  result = {}
  for key,val in number1.items():
    if isinstance(val,list):
      # val = val[::-1]
      val = val[0:5:-1]
      result[key] = val
    else:
      result[key] = val
  return result

print(number1_func())
```
