```python
data = [10, "Ikan", 7, 8, "Bunga", "Buah"]  # Menginisialisasi list data dengan beberapa nilai.
indeks = 0  # Menginisialisasi variabel indeks sebagai penanda posisi saat ini dalam iterasi.

for elemen in data:  # Melakukan iterasi melalui setiap elemen dalam list data.
  indeks = indeks + 1  # Meningkatkan nilai indeks setiap kali iterasi untuk menunjukkan posisi elemen saat ini.
  print("element ke ", indeks, "=", elemen)  # Mencetak indeks dan nilai elemen saat ini.
```

```python
data = [10,9,7,8,10,8]
indeks = 0

while indeks<6:
  if data[indeks]>=10:
    print("Element ke-", indeks, "=", data[indeks]+5)
  else:
    print("Element ke-", indeks, "=", data[indeks]+10)
  indeks = indeks +1
```
